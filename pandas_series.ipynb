```python
"""
创建一个包含10名学生数学成绩的Series，成绩范围在50-100之间。计算平均分、最高分、最低分，并找出高于平均分的学生人数。
"""
import pandas as pd
import numpy as np
np.random.seed(0)
scores = pd.Series(np.random.randint(50,101,10),index=('学生'+str(i) for i in range(1,11)))
print(scores)
print('平均分为',scores.mean())
print('最高分为',scores.max())
print('最低分为',scores.min())
print('高于平均分的学生人数',np.size(np.where(scores>scores.mean())))
```

```python
"""
给定某城市一周每天的最高温度Series，完成以下任务：
找出温度超过30度的天数
计算平均温度
将温度从高到低排序
找出温度变化最大的两天
"""
temp = pd.Series([29, 29, 30, 32, 33, 29, 35],index=['周一', '周二', '周三', '周四', '周五', '周六', '周日'])
print("温度超过30度的天数",temp[temp > 30].count())
print(f"平均温度是{temp.mean()}度")
print("从高到低排序\n",temp.sort_values(ascending=False))
temp_diff = temp.diff().abs()
max_diff_days = temp_diff.nlargest(2).index.tolist()
print(f"温度变化最大的两天: {max_diff_days}")
```
